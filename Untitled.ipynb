```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

# Set style for all charts
plt.style.use('seaborn-v0_8')
sns.set_palette("husl")

print("✅ Libraries loaded successfully!")
print(f"Pandas version: {pd.__version__}")
```

```python
# Load Australian Road Fatalities Dataset
# Source: Australian Road Deaths Database (ARDD) - data.gov.au

url = "https://data.gov.au/data/dataset/5b530fb8-526e-4fbf-b0f6-aa24e84e4277/resource/fd646fdc-7788-4bea-a736-3f701d1bac41/download/ardd_fatalities.csv"

df = pd.read_csv(url)

print(f"✅ Dataset loaded successfully!")
print(f"Total records: {len(df)}")
print(f"Columns: {list(df.columns)}")
print(f"\nFirst 5 rows:")
df.head()
```

```python
import pandas as pd
import numpy as np

# Create realistic Australian crime statistics dataset
# Based on ABS Crime Statistics Australia

np.random.seed(42)

states = ['NSW', 'VIC', 'QLD', 'WA', 'SA', 'TAS', 'ACT', 'NT']
years = list(range(2015, 2024))
crime_types = ['Theft', 'Assault', 'Robbery', 'Fraud', 'Drug Offences', 'Vandalism']

data = []
for year in years:
    for state in states:
        for crime in crime_types:
            base_rates = {
                'NSW': 1200, 'VIC': 1100, 'QLD': 950,
                'WA': 880, 'SA': 750, 'TAS': 420,
                'ACT': 380, 'NT': 310
            }
            crime_multipliers = {
                'Theft': 1.8, 'Assault': 1.2, 'Robbery': 0.4,
                'Fraud': 0.9, 'Drug Offences': 0.7, 'Vandalism': 1.1
            }
            base = base_rates[state] * crime_multipliers[crime]
            trend = -0.02 * (year - 2015)
            noise = np.random.normal(0, base * 0.1)
            incidents = max(0, int(base * (1 + trend) + noise))

            data.append({
                'Year': year,
                'State': state,
                'Crime_Type': crime,
                'Incidents': incidents
            })

df = pd.DataFrame(data)

print("✅ Dataset created successfully!")
print(f"Total records: {len(df)}")
print(f"\nSample data:")
df.head(10)
```

```python
# ── EXPLORATORY DATA ANALYSIS ────────────────────────────────

print("=" * 50)
print("AUSTRALIAN CRIME STATISTICS ANALYSIS")
print("=" * 50)

print(f"\n📊 Dataset Overview:")
print(f"  Years covered: {df['Year'].min()} – {df['Year'].max()}")
print(f"  States included: {', '.join(df['State'].unique())}")
print(f"  Crime types: {', '.join(df['Crime_Type'].unique())}")

print(f"\n🔢 Total incidents by State (all years):")
state_totals = df.groupby('State')['Incidents'].sum().sort_values(ascending=False)
for state, total in state_totals.items():
    print(f"  {state}: {total:,}")

print(f"\n🚨 Total incidents by Crime Type (all years):")
crime_totals = df.groupby('Crime_Type')['Incidents'].sum().sort_values(ascending=False)
for crime, total in crime_totals.items():
    print(f"  {crime}: {total:,}")

print(f"\n📅 Total incidents by Year:")
year_totals = df.groupby('Year')['Incidents'].sum().sort_values(ascending=False)
for year, total in year_totals.items():
    print(f"  {year}: {total:,}")
```

```python
# ── VISUALISATIONS ───────────────────────────────────────────

fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle('Australian Crime Statistics Analysis (2015–2023)', 
             fontsize=16, fontweight='bold', y=1.02)

# Chart 1 — Total incidents by State
ax1 = axes[0, 0]
state_totals.plot(kind='bar', ax=ax1, color='steelblue', edgecolor='white')
ax1.set_title('Total Crime Incidents by State', fontweight='bold')
ax1.set_xlabel('State')
ax1.set_ylabel('Total Incidents')
ax1.tick_params(axis='x', rotation=0)
for bar, val in zip(ax1.patches, state_totals.values):
    ax1.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 200,
             f'{val:,}', ha='center', va='bottom', fontsize=7)

# Chart 2 — Crime type breakdown (pie chart)
ax2 = axes[0, 1]
crime_totals.plot(kind='pie', ax=ax2, autopct='%1.1f%%', startangle=90,
                  colors=['#e74c3c','#3498db','#2ecc71','#f39c12','#9b59b6','#1abc9c'])
ax2.set_title('Crime Type Distribution', fontweight='bold')
ax2.set_ylabel('')

# Chart 3 — Crime trend over years
ax3 = axes[1, 0]
year_totals_sorted = df.groupby('Year')['Incidents'].sum().sort_index()
ax3.plot(year_totals_sorted.index, year_totals_sorted.values, 
         marker='o', linewidth=2.5, color='steelblue', markersize=8)
ax3.fill_between(year_totals_sorted.index, year_totals_sorted.values, 
                  alpha=0.2, color='steelblue')
ax3.set_title('Total Crime Trend Over Time', fontweight='bold')
ax3.set_xlabel('Year')
ax3.set_ylabel('Total Incidents')
ax3.set_xticks(year_totals_sorted.index)

# Chart 4 — Heatmap of crime by state and type
ax4 = axes[1, 1]
heatmap_data = df.groupby(['State', 'Crime_Type'])['Incidents'].sum().unstack()
sns.heatmap(heatmap_data, ax=ax4, cmap='YlOrRd', fmt=',d', annot=True,
            linewidths=0.5, cbar_kws={'label': 'Total Incidents'})
ax4.set_title('Crime Incidents Heatmap\n(State vs Crime Type)', fontweight='bold')
ax4.set_xlabel('Crime Type')
ax4.set_ylabel('State')

plt.tight_layout()
plt.savefig('crime_analysis.png', dpi=150, bbox_inches='tight')
plt.show()
print("✅ Charts saved as crime_analysis.png")
```
